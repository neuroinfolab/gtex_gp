# GTEx / AHBA Coordinate Overlay in 3D MNI Space

This notebook renders interactive Plotly overlays of AHBA target parcel coordinates with GTEx coordinates under different spatial preprocessing and representative-point choices.

It supports:

- native or left-hemisphere-mirrored preprocessing
- centroid or medoid representative-point assignment
- visualization of either the representative GTEx points used for parcel matching or the full transformed coordinate sets

AHBA parcels are shown in neutral gray. GTEx points are colored with a discrete qualitative palette by their mapped AHBA parcel group.

## GTEx Tissue vs AHBA Parcel Genetic Similarity

This section computes a simple data-driven correspondence table by averaging expression within each GTEx tissue and within each AHBA parcel, then ranking AHBA parcels for each GTEx tissue by Pearson correlation.

```python
META_COLS = ["subject", "age", "sex", "dataset", "tissue_or_parcel", "coordinates"]
gene_cols = [c for c in gxp_samples_df.columns if c not in META_COLS]
# Subset to 100 genes in gtex_gp/data/raw/ahba_100hvg.txt if it exists

GENE_SUBSET_PATH = "./data/raw/ahba_100hvg.txt"
if os.path.exists(GENE_SUBSET_PATH):
    with open(GENE_SUBSET_PATH) as f:
        selected_genes = [line.strip() for line in f if line.strip()]
    selected_genes_set = set(selected_genes)
    gene_cols = [g for g in gene_cols if g in selected_genes_set]
    print(f"Subsetting to {len(gene_cols)} genes from {GENE_SUBSET_PATH}")
else:
    print("No gene subset file found, using all genes.")

gtex_mean_by_tissue = (
    gxp_samples_df.loc[gxp_samples_df["dataset"].astype(str) == "GTEx"]
    .groupby("tissue_or_parcel")[gene_cols]
    .mean(numeric_only=True)
)

ahba_mean_by_parcel = (
    gxp_samples_df.loc[gxp_samples_df["dataset"].astype(str) == "AHBA"]
    .groupby("tissue_or_parcel")[gene_cols]
    .mean(numeric_only=True)
)

overlay_data = load_coordinate_overlay_data(csv_path=CSV_PATH)
ahba_coords = overlay_data["ahba_parcels"][["tissue_or_parcel", "coord_x", "coord_y", "coord_z"]].copy()
ahba_coord_lookup = {
    row["tissue_or_parcel"]: (round(float(row["coord_x"]), 2), round(float(row["coord_y"]), 2), round(float(row["coord_z"]), 2))
    for _, row in ahba_coords.iterrows()
}

def pearson_safe(x, y):
    x = pd.to_numeric(pd.Series(x), errors="coerce").to_numpy(dtype=float)
    y = pd.to_numeric(pd.Series(y), errors="coerce").to_numpy(dtype=float)
    mask = np.isfinite(x) & np.isfinite(y)
    if mask.sum() < 2:
        return np.nan
    x = x[mask]
    y = y[mask]
    if np.std(x) < 1e-12 or np.std(y) < 1e-12:
        return np.nan
    return float(np.corrcoef(x, y)[0, 1])

TOP_K = 20
ranked_rows = []
for tissue_name, tissue_vec in gtex_mean_by_tissue.iterrows():
    pairs = []
    for parcel_name, parcel_vec in ahba_mean_by_parcel.iterrows():
        r = pearson_safe(tissue_vec, parcel_vec)
        coords = ahba_coord_lookup.get(parcel_name)
        pairs.append((parcel_name, r, coords))
    pairs = sorted(pairs, key=lambda x: (-np.inf if not np.isfinite(x[1]) else -x[1], x[0]))
    row = {"source_gtex_tissue": tissue_name}
    for rank, (parcel_name, r, coords) in enumerate(pairs[:TOP_K], start=1):
        row[f"match_{rank}"] = (parcel_name, None if not np.isfinite(r) else round(float(r), 4), coords)
    ranked_rows.append(row)

tissue_to_ahba_ranked_df = pd.DataFrame(ranked_rows).sort_values("source_gtex_tissue").reset_index(drop=True)
tissue_to_ahba_ranked_df
```
