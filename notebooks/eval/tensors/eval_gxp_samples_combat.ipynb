# GTEx + AHBA ComBat Tensors — T3 styling

Recreates the ComBat-harmonization tensor plots (raw-matched & harmonized, standalone + joint),
styled to match **Tensor 3** of `eval_gxp_samples_pipeline.ipynb`: shared `BASE`/`BIG_FONTS`,
`BOX_ZOOM=1.0`, region ticks every 2 (no region labels) + region y-axis title, subject ticks
every 2, gene ticks every 5, harmonized joint on a shared color scale.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

from src.eval_utils.eval_style import set_academic_style
from src.eval_utils.eda_core import prepare_pre_post_harmonization_cached
from src.eval_utils.results_eda import EDAConfig
from src.eval_utils.eval_samples import (
    MatchingContext, build_combat_tensor_view, build_joint_tensor_view,
    plot_gtex_tensor_voxels, plot_joint_tensor_voxels,
)

set_academic_style()
```

## 0 · Config (matches Tensor-3 defaults)

```python
CSV_PATH   = 'data/raw/gxp_samples.csv'
GENE_LIST  = 'richiardi2015'
N_GENES    = 100
GENE_TICK_STEP = 5

MATCHING_POLICY            = 'centroids_and_volumes'
MATCHING_POLICY_HEMI_MODE  = 'force_left'
GTEX_REP_MODE              = 'centroid'
GTEX_HEMI_MODE             = 'mirror_left'
COLLAPSE_CEREBELLUM        = False
MIN_OBSERVED_PARCELS       = 5
COMBAT_USE_COVARIATES      = True
RANDOM_SEED                = 42

N_GTEX_SUBJECTS = 75
N_AHBA_SUBJECTS = 6
N_REGIONS       = 74
REGION_TICK_STEP  = 2
SUBJECT_TICK_STEP = 4
STACK_ORDER       = ('AHBA', 'GTEx')

BOX_ZOOM = 1.0
BIG_FONTS = {'title': 'xxl', 'xlabel': 'xl', 'ylabel': 'xl', 'zlabel': 'xl',
             'cbar_label': 'xl', 'xtick': 'm', 'ytick': 'm', 'ztick': 'm'}

BASE = dict(
    missing_rgba=(0.78, 0.78, 0.78, 1.0),
    future_imputation_rgba=(0.78, 0.78, 0.78, 0.3),
    figsize='auto', dpi=300, axes_bbox='auto', colorbar_bbox='auto',
    box_aspect='compressed_data', box_zoom=BOX_ZOOM, title_pad=2.0,
    axis_assignment=('gene', 'region', 'subject'), elev=20.0, azim=24.0,
    gene_tick_step=GENE_TICK_STEP, tick_label_pad=0.0,
    subject_label_mode='full', gene_label_mode='full',
    show_x_axis_label=True, show_axis_labels=False,
    show_gene_ticklabels=False, show_region_label_colors=False,
    show_axis_lines=True, show_tick_lines=True, show_grid=False, show_legend=False,
    mask_render_mode='two_pass', font_sizes=BIG_FONTS,
)

# T3-style tick/label overrides shared by every plot here.
T3 = dict(subject_tick_step=SUBJECT_TICK_STEP, region_tick_step=REGION_TICK_STEP,
          show_subject_ticklabels=True, show_region_ticklabels=False, show_y_axis_label=True)
```

```python
cfg = EDAConfig(
    csv_path=CSV_PATH, gene_scope='allgenes',
    min_observed_parcels=MIN_OBSERVED_PARCELS, combat_use_covariates=COMBAT_USE_COVARIATES,
    gtex_rep_mode=GTEX_REP_MODE, gtex_hemi_mode=GTEX_HEMI_MODE,
    matching_policy=MATCHING_POLICY, matching_policy_hemi_mode=MATCHING_POLICY_HEMI_MODE,
    collapse_cerebellum=COLLAPSE_CEREBELLUM,
)
prepost = prepare_pre_post_harmonization_cached(cfg)
ctx = MatchingContext.from_cfg(cfg, prepost)
print(f'parcels={len(ctx.regions_full)}  genes={len(ctx.genes_full)}')
```

## 1 · Build raw-matched & harmonized views (both datasets)

```python
_cmb = dict(cfg=cfg, matching_context=ctx, region_ordering='region_matched_superset',
            gene_panel=GENE_LIST, n_regions=N_REGIONS, n_genes=N_GENES, random_seed=RANDOM_SEED)

gtex_rm = build_combat_tensor_view(prepost, dataset='GTEx', stage='raw_matched', n_subjects=N_GTEX_SUBJECTS, **_cmb)
gtex_h  = build_combat_tensor_view(prepost, dataset='GTEx', stage='harmonized', n_subjects=N_GTEX_SUBJECTS, **_cmb)
ahba_rm = build_combat_tensor_view(prepost, dataset='AHBA', stage='raw_matched', n_subjects=N_AHBA_SUBJECTS, **_cmb)
ahba_h  = build_combat_tensor_view(prepost, dataset='AHBA', stage='harmonized', n_subjects=N_AHBA_SUBJECTS, **_cmb)
for nm, v in [('gtex_rm', gtex_rm), ('gtex_h', gtex_h), ('ahba_rm', ahba_rm), ('ahba_h', ahba_h)]:
    print(f'{nm}: {v.values.shape}  stage={v.pipeline_stage!r}')
```

## 2 · Standalone tensors (pre/post ComBat)

```python
def render_single(view, title):
    ns, nr, ng = view.values.shape
    fig, ax = plot_gtex_tensor_voxels(view, **{**BASE, **T3})
    ax.set_title(f'{title}  ({ns} subj x {nr} reg x {ng} genes)', fontsize=18, pad=2.0)
    plt.show()
```

```python
render_single(gtex_rm, 'GTEx raw-matched (pre-ComBat)')
```

```python
render_single(gtex_h,  'GTEx harmonized (post-ComBat)')
```

```python
render_single(ahba_rm, 'AHBA raw-matched (pre-ComBat)')
```

```python
render_single(ahba_h,  'AHBA harmonized (post-ComBat)')
```

## 3 · Joint tensors (stacked GTEx + AHBA)

```python
raw_matched_joint = build_joint_tensor_view(gtex_rm, ahba_rm, stack_order=STACK_ORDER)
harmonized_joint  = build_joint_tensor_view(gtex_h,  ahba_h,  stack_order=STACK_ORDER)
print('raw_matched_joint', raw_matched_joint.values.shape, '| harmonized_joint', harmonized_joint.values.shape)
```

### Raw-matched joint (pre-ComBat — per-dataset color scale)

```python
fig, ax = plot_joint_tensor_voxels(
    raw_matched_joint, **{**BASE, **T3, 'dataset_gap': 1,
        'normalization': 'per_dataset_minmax', 'cbar_label': 'expression'},
)
ax.set_title('GTEx + AHBA — raw-matched (pre-ComBat)', fontsize=18, pad=2.0)
plt.show()
```

### Harmonized joint (post-ComBat — shared color scale, == Tensor 3)

```python
fig, ax = plot_joint_tensor_voxels(
    harmonized_joint, **{**BASE, **T3, 'dataset_gap': 1,
        'normalization': 'shared', 'cbar_label': 'harmonized expression'},
)
ax.set_title('GTEx + AHBA — ComBat harmonized', fontsize=18, pad=2.0)
plt.show()
```
